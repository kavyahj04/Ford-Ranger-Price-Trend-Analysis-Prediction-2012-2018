```python
#Import Modules
import numpy as np
import matplotlib.pyplot as plt
```

```python

# ---------------------------------------------
# Load observed data

# Note: Years 2012–2018 are missing; these will be predicted later
# ---------------------------------------------


years_input = np.array([2000, 2001, 2002, 2003, 2004, 2005, 2006, 2007, 2008, 2009, 2010, 2011, 2019, 2020, 2021, 2022, 2023, 2024, 2025], dtype = int)
min_price_input = np.array([11580, 11960, 12565, 13645, 14575, 14610,  14450, 13970,  14490,  16395,  17820,  18160,  24000,  24110,  24820,  25980,  27400,  32820,  33350], dtype = float)
max_price_input = np.array([19785, 24335, 25010, 25450, 26015, 26795,  26670, 24425,  24350,  25805,  25800,  26070,  38565,  38675,  39035,  39730,  40945,  55720,  56070], dtype = float)
```

```python


# -----------------------------------------------
# Initialize parameters
# ------------------------------------------------
x = years_input
y = min_price_input

# Initialize theta0 (intercept) and theta1 (slope) to small values
theta0 = 0.01
theta1 = 0.02

# alpha = learning rate (step size in gradient descent)
# iterations = number of times to update parameters
# m = number of training samples
alpha = 0.00000001
iterations = 100
losses_min_price = []
m = x.size

#Apply General Algothim to compute new theta0 and theta1 values along with objective function J
for i in range(1, iterations + 1):

  #compute output f(x)
  f_of_x = theta0 + theta1 * x

  #compute error
  error = f_of_x - y

  #compute gradient dj/d_theta0 and dj/d_theta1
  grad_theta0 = (1.0 / m) * np.sum(error)
  grad_theta1 = (1.0 / m) * np.sum(error * x)

  #compute new theta0 and theta1 values
  theta0 = theta0 - alpha * grad_theta0
  theta1 = theta1 - alpha * grad_theta1

  #compute new f_of_x
  f_of_x_new = theta0 + theta1 * x

  #compute objective function
  J = (1.0 / (2.0 * m)) * np.sum((f_of_x_new - y) ** 2)
  losses_min_price.append(J)

theta0_min_base = round(theta0, 6)
theta1_min_base = round(theta1, 6)
losses_min_base = np.array(losses_min_price)

#final theta0 and theta1 values after 100 iterations
print(f"Final theta: theta0={theta0:.6f}, theta1={theta1:.6f}")




```

```python
#plotted loss curve for 100 iterations with x label as Iterations and y label as Obtained Loss

# X-axis: iteration numbers from 1 to total number of iterations
plt.plot(range(1, iterations + 1), losses_min_base)

# Label for the X-axis
plt.xlabel("Iteration")

# Label for the Y-axis with the formula of the cost function
plt.ylabel("Loss J = (1/(2m)) Σ (pred − y)^2")

# Title of the plot
plt.title("MinPrice — Loss Curve")

# Add grid lines to make the plot easier to read
plt.grid(True)

# Display the plot
plt.show()
```

```python
#to check initial and final obtained objective functions
print("first J:", losses_min_base[0], " last J:", losses_min_base[-1])
```

```python
#Predict missing data (2012–2018)
missing_years = np.array([2012, 2013, 2014, 2015, 2016, 2017, 2018], dtype=float)
```

```python

#Apply linear model f_theta(x) = theta0 + theta1 * x
pred_min_missing_base = theta0_min_base + theta1_min_base * missing_years

#Print year–price pairs
for year, price in zip(missing_years.astype(int), pred_min_missing_base):
    print(f"Year {year}: Predicted Min Price = ${price:.2f}")
```

```python


#Standardize input feature (year) to improve convergence speed
#Formula: x' = (x - μ) / σ
#μ = mean of years, σ = standard deviation of years

mean = years_input.mean()
sigma = years_input.std()

# Apply scaling to training years
years_input_scaled = (years_input - mean) / sigma

# Apply the same scaling to missing years (2012–2018)
missing_years_scaled = (missing_years - mean) / sigma

# Print scaling parameters
print(f"Mean of years (μ): {mean}")
print(f"Standard deviation of years (σ): {sigma}")
print(f"Scaled years dtype: {years_input_scaled.dtype}")
```

```python
# -----------------------------------------------
# Initialize parameters
# -----------------------------------------------
x = years_input_scaled
y = min_price_input

# Initialize theta0 (intercept) and theta1 (slope) to small values
theta0, theta1 = 0.1, 0.2

# iterations = number of times to update parameters
# m = number of training samples
iterations = 100
iters = 100
losses_min_scaled = []
m = years_input_scaled.size

#dynamic learning rat function
def alpha_dlr(t, c=0.5, tau=1.0):
    return c / (t + tau)

#Apply General Algothim to compute new theta0 and theta1 values along with objective function J
for i in range(1, iterations + 1):
  #compute output f(x)
  f_of_x = theta0 + theta1 * x

  #compute error
  error = f_of_x - y

  #compute gradient dj/d_theta0 and dj/d_theta1
  grad_theta0 = (1.0 / m) * np.sum(error)
  grad_theta1 = (1.0 / m) * np.sum(error * x)

  #compute synamc learning rate
  new_alpha = alpha_dlr(i)

  #compute new theta0 and theta1 values
  theta0 = theta0 - new_alpha * grad_theta0
  theta1 = theta1 - new_alpha * grad_theta1

  #compute new f_of_x
  f_of_x_new_2 = theta0 + theta1 * x

  #compute objective function
  J = (1.0 / (2.0 * m)) * np.sum((f_of_x_new_2 - y) ** 2)
  losses_min_scaled.append(J)

theta0_min_base_scaled = round(theta0, 6)
theta1_min_base_scaled = round(theta1, 6)
losses_min_base_scaled = np.array(losses_min_scaled)

#final theta0 and theta1 values after 100 iterations
print(f"Final theta: theta0={theta0:.6f}, theta1={theta1:.6f}")
```

```python
# Plot the loss curve for the minimum price model after feature scaling

# X-axis: iteration numbers from 1 to total number of iterations
plt.plot(range(1, iterations + 1), losses_min_base_scaled)

# Label for the X-axis
plt.xlabel("Iteration")

# Label for the Y-axis with the cost function formula
plt.ylabel("Loss J = (1/(2m)) Σ (pred − y)^2")

# Title of the plot (indicating it's MinPrice loss curve after scaling, 100 iterations)
plt.title("MinPrice — Loss (100 iters) after scaling")

# Add grid lines for easier visualization
plt.grid(True)

# Display the plot
plt.show()
```

***Formula of the gradient of the loss w.r.t. the theta or weights***

Cost function:
J(theta0, theta1) = (1 / (2 * m)) * Σ (f_theta(xi) - yi)^2
where f_theta(x) = theta0 + theta1 * x

Gradients:
∂J/∂theta0 = (1 / m) * Σ (f_theta(xi) - yi)
∂J/∂theta1 = (1 / m) * Σ (f_theta(xi) - yi) * xi
